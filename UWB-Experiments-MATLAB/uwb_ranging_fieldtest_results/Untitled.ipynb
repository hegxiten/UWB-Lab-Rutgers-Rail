```python
import numpy as np
import matplotlib.pyplot as plt

#generate some data
data = np.random.normal(size=100)

#define the plot
fig, ax = plt.subplots()

#plot the data as a histogram
ax.hist(data, orientation=u'horizontal')

#invert the order of x-axis values
ax.set_xlim(ax.get_xlim()[::-1])

#move ticks to the right
ax.yaxis.tick_right()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.arange(0.01, 5.0, 0.01)
s1 = np.sin(2 * np.pi * t)
s2 = np.exp(-t)
s3 = np.sin(4 * np.pi * t)

ax1 = plt.subplot(121)
plt.plot(t, s1)
plt.tick_params('x', labelsize=6)

# share x only
ax2 = plt.subplot(122, sharey=ax1)
plt.plot(t, s2)
# make these tick labels invisible
plt.tick_params('x', labelbottom=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt

#generate some data
data = np.random.normal(size=100)

fig, axs = plt.subplots(1,2,sharey=True)
fig.subplots_adjust(wspace=0)
axs[1].hist(data, orientation=u'horizontal')

#move ticks to the right
axs[1].yaxis.tick_right()

#define the plot
axs[0].hist(data, orientation=u'horizontal')

#invert the order of x-axis values
axs[0].set_xlim(axs[1].get_xlim()[::-1])

#move ticks to the right
axs[0].yaxis.tick_right()

plt.show()
```
